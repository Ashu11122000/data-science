```python
import matplotlib.pyplot as plt

category_colors = {"Furniture": 0, "Office Supplies": 1, "Technology": 2}

colors = df_clean["Category"].map(category_colors)

plt.scatter(df_clean["Sales"], df_clean["Profit"], c=colors)

plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

category_colors = {"Furniture": 0, "Office Supplies": 1, "Technology": 2}

colors = df_clean["Category"].map(category_colors)

plt.figure(figsize=(10, 6))

plt.scatter(df_clean["Sales"], df_clean["Profit"], c=colors, alpha=0.7)

plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit by Product Category")

legend_elements = [
    Line2D(
        [0], [0],
        marker="o",
        linestyle="",
        label=category,
        markerfacecolor=plt.cm.viridis(color_value / 2),
        markeredgecolor="none"
    )
    for category, color_value in category_colors.items()
]

plt.legend(handles=legend_elements, title="Product Category")

plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

category_colors = {
    "Furniture": 0,
    "Office Supplies": 1,
    "Technology": 2
}

colors = df_clean["Category"].map(category_colors)

plt.figure(figsize=(10, 6))

plt.scatter(
    df_clean["Sales"],
    df_clean["Profit"],
    c=colors,
    alpha=0.7
)

plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Sales vs Profit by Product Category")

legend_elements = [
    Line2D(
        [0], [0],
        marker="o",
        linestyle="",
        label=category,
        markerfacecolor=plt.cm.viridis(color_value / 2),
        markeredgecolor="none"
    )
    for category, color_value in category_colors.items()
]

plt.legend(handles=legend_elements, title="Product Category")

plt.savefig(
    "../outputs/figures/sales_profit_scatter.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
