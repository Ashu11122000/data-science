```python
df_clean = df.copy()

df_clean.loc[0, "Profit"] = None

print("Original rows:", len(df))
print("Missing Profit values:", df_clean["Profit"].isnull().sum())
```

```python
df_clean["Profit"] = df_clean["Profit"].fillna(df_clean["Profit"].median())

print("Missing Profit values:", df_clean["Profit"].isnull().sum())
```

```python
df_clean["Profit_Margin"] = df_clean["Profit"] / df_clean["Sales"]

df_clean[["Sales", "Profit", "Profit_Margin"]].head()
```

```python
df_clean.groupby("Region")["Sales"].sum()
```

```python
df_clean.to_csv("../data/processed/cleaned_sales_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
sales_by_region = df_clean.groupby("Region")["Sales"].sum()

sales_by_region.to_csv("../outputs/tables/sales_by_region.csv")

print("Sales by region saved successfully!")
```
